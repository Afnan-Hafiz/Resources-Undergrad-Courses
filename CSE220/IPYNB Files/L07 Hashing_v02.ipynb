### Hash Table or Hash Map:
- Use **key-value pairs** to store data instead of storing against an index.
- The idea is to find the data in **constant time** using that **key.**


- Example:
   - Python dictionary.
   - JSON (JavaScript Object Notation)

### Why use hashing?
- **Fast Data Retrieval (Time Complexity: O(1))**
   - Hashing allows for constant-time complexity, 𝑂(1), for search, insertion, and deletion in an ideal scenario.
    - A data structure that allows for the very fast retrieval of data no matter how much data there is.
    - Hashing is an example of a **space-time tradeoff**.
        - If memory is infinite, the entire key can be used directly as an index to locate its value with a single memory access.
- **Efficient Handling of Large Datasets**
  - Hashing provides a way to manage large datasets effectively by distributing the data evenly across a fixed-size table.
  - This ensures that the data remains accessible without searching through an entire dataset.
  
- **Avoiding Sequential Search**
   - Unlike linear search (O(n)) or binary search (O(log n)), hashing avoids scanning the entire dataset, significantly speeding up operations.

### Hash table:
The array where both key and values are stored.

### Hashing Algorithm/Hashing Function:
It represents the **calculations applied to a key** which may be a very large **number** or a very **large string** to transform it into a **relatively small index** that corresponds to a position in a hash table. The output, typically **a number**, is called the **hash code or hash value.**

- This **index number** is effectively a **memory address**.
- There is a lot of Hashing Algorithms to choose from.

#### Objectives of hash function
- Minimize collision
- Uniform distribution of hash values
- Easy to calculate
- Resolve any collisions

## Collisions
Sometimes a hash function applied to **different keys** could generate the **same index number** for both. However, both items cannot be stored in the same place.

### Solutions
1) Open addressing (Solution of collision removal: Linear probing, double hashing) -- Optional

2) Closed addressing (Solution of collision removal: Forward chaining)


### Open addressing: ( Optional, Not needed)
Resolving a collision by placing an item somewhere other than its calculated address is called open address because every location is opened to any other item.
- Open addressing can use a variety of techniques to decide where to place an item that does not go where it should.

## Problem with linear probing
- To lookup an item in the hash table, hashing function is applied again.
- If there is a collision and the item is not in the calculated position,
   - then finding an item would also involve **linear probing,**
   - that is a **linear search.**
- Increases chance of collision
***We do not use linear probing due to this issue. Forward chaining solves this issue. One of the popular methods***

### Double hashing: (Optional, Not needed)
 - Better than linear probing.
 - Use another hash function if we find any collision.
    - The first hash function determines the key and
    - the second hash function determines the jump from the given key if collision occurs.

### Load factor
In hashing, the load factor is a measure of **how full a hash table is.**
- Load factor = number of items / total number of slots

For the example below,
- slots are 5, pairs/items are 7.
- So, load factor = 7/5 = 1.4.
- Ideal is to have 1. (No collisions)

### Collion removal process 02: Forward chaining
#### Forward Chaining:
- The idea of Forward Chaining is to **store a linked list in the index**.
- Forward chaining is a simple and effective way to handle collisions in a hash table, and it ensures that all key-value pairs are retained **without overwriting or losing any data.**

Operations:
- INSERTION: As a result, whenever a collision occurs, the value is stored at the **beginning of the linked list** in that particular index where collision happens.
   - New elements/inserted values are stored at the beginning of the linked list to decrease the insertion time. **O(1) - constant time**
- SEARCH: Whenever we search for an item, it will **traverse the list** and check the key to find the value.
    - search time is longer if collisions occur frequently.
    
- When k keys have been hashed into the same slot, **deletion and search time is Θ(k)** which is **linear time (Length of the list for that index)**
    - For this reason, choosing the appropriate hash function to minimize the collision is very important.

```python
class Node:
    def __init__(self, key, value=None, next=None):
        self.key = key
        self.value = value
        self.next = next

#############################################################

class HashTable_Example:
    def __init__(self, size=5):
        self.size = size
        self.table = [None] * self.size  # Initialize an empty hash table with the specified size

    def hash_function(self, key):
        """Simple hash function."""
        hash_value = 0
        for char in key:
            hash_value += ord(char)
        return hash_value % self.size

    def insert(self, key, value):
        """Insert a key-value pair into the hash table."""
        index = self.hash_function(key)

        # start creating the chain for the 1st element
        if self.table[index] is None:
            self.table[index] = Node(None)  # Dummy head node

        # prepend after dummy head
        newNode = Node(key, value)

        predecessor = self.table[index] # predecessor is dummy head
        # print("predecessor",predecessor.key, predecessor.value)
        successor = predecessor.next

        newNode.next = successor  # Link the successor
        predecessor.next = newNode  # Link the predecessor


    def search(self, key):
        """Search for a key in the hash table."""
        index = self.hash_function(key)
        current = self.table[index].next # starting after dummy head

        while current!=None: # going till tail
            if current.key == key:  # If key matches, return True
                return True
            current = current.next  # Move to the next node

        return False  # If the key is not found


    def remove(self, rem_key):
        """Remove a key from the hash table using predecessor, successor, and remNode."""
        index = self.hash_function(rem_key)

        # Start at the dummy head
        if self.table[index] is None:
            print(f"Key '{rem_key}' not found.")
            return

        predecessor = self.table[index]  # Predecessor (head node), i = 0
        remNode = predecessor.next  # Node to be removed (remNode), j = i+1

        # Traverse the chain to find the remNode
        while remNode is not None:
            if remNode.key == rem_key:  # Found the node to remove
                successor = remNode.next  # The node after remNode
                predecessor.next = successor  # Bypass remNode
                print(f"Removed ({remNode.key}, {remNode.value})")
                return
            predecessor = remNode  # Move predecessor forward, i++
            remNode = remNode.next  # Move remNode forward, j++

        print(f"Key '{rem_key}' not found.")


    def print_hash_table(self):
        """Print the hash table."""
        i = 0
        for node in self.table: # array slots
            print(f"{i}:", end=" ")
            current = node
            while current !=None and current.next!=None:
                print(f"({current.next.key}, {current.next.value})", end=" → ")
                current = current.next
            print("None")  # End of the linked list
            i += 1
        print()
```

#### Coding style-01 (Optional)

```python
class Node:
    """Node class for singly linked list."""
    def __init__(self, key=None, value=None, next=None):
        """Initialize a node."""
        self.key = key
        self.value = value
        self.next = next
#############################################################

class SinglyLinkedList:
    """Singly linked list with a dummy head for collision chains."""
    def __init__(self):
        self.head = Node()  # Dummy head

    def find(self, key):
        """Find a node by key."""
        current = self.head.next
        while current!= None:
            if current.key == key:
                return current
            current = current.next
        return None

    def prepend(self, key, value):
        """Prepend a new key-value pair to the list."""
        node = self.find(key)
        if node is not None: # Update value if key exists
            node.value = value
        else: # Prepend code
            new_node = Node(key, value)
            new_node.next = self.head.next
            self.head.next = new_node

    def delete(self, key):
        """Delete a node by key using successor, predecessor, and rem_node."""
        predecessor = self.head
        rem_node = predecessor.next  # Node to be removed

        while rem_node!= None:  # Traverse the list
            if rem_node.key == key:
                successor = rem_node.next  # Next node in the list
                predecessor.next = successor  # Bypass rem_node
                return True  # Deletion successful
            predecessor = rem_node  # Move forward
            rem_node = rem_node.next  # Update rem_node

        return False  # Key not found

    def display(self):
        """Display all key-value pairs in the list."""
        current = self.head.next
        result = []
        while current!= None:
            result.append((current.key, current.value))
            current = current.next
        return result
```

```python
class HashTable:
    """Hash table with separate chaining using singly linked lists."""
    def __init__(self, size):
        self.size = size
        self.table = [SinglyLinkedList() for _ in range(size)]

    def _hash(self, key):
        """Simple hash function."""
        hash_value = 0
        for char in key:
            hash_value += ord(char)
        return hash_value % self.size

    def insert(self, key, value):
        """Insert or update a key-value pair in the hash table."""
        index = self._hash(key)
        self.table[index].prepend(key, value)

    def search(self, key):
        """Search for a value by key."""
        index = self._hash(key)
        node = self.table[index].find(key)
        if node != None:
            return node.value
        else:
            return None

    def delete(self, key):
        """Delete a key-value pair."""
        index = self._hash(key)
        return self.table[index].delete(key)

    def display(self):
        """Display the entire hash table."""
        index = 0
        for chain in self.table:
            print(f"Bucket {index}: {chain.display()}")
            index += 1
```

### Cryptographic Hashing/one-way hashing (Optional)
- Fundamental concept in computer science and cryptography.
- It is a mathematical function that takes an input (or "message") of arbitrary size and produces a fixed-size string of characters,
    - typically represented in hexadecimal format.
- This fixed-size output is often referred to as the hash value, hash code, or simply hash.

#### Key properties of cryptographic hashing functions are:

1. **Deterministic:** For a given input, the hash function always produces the same output. This property is crucial for consistency and predictability.


2. **Irreversibility (One-way):** The hash function is designed to be irreversible, meaning it should be computationally infeasible to derive the original input from its hash value. This property ensures that the original data remains secure even if the hash is known.

3. **Collision resistance:** It should be extremely difficult to find two different inputs that produce the same hash value. In other words, it should be computationally infeasible to create a collision intentionally.





#### Applications of Cryptographic hashing
Numerous applications in information security and computer science, including but not limited to:

- **Password storage:** Storing passwords securely by hashing them, preventing direct exposure of user passwords in databases.

- **Data integrity verification:** Verifying that data has not been tampered with by comparing the hash of the original data with the computed hash of the received data.

- **Digital signatures:** Used in digital signature algorithms to ensure authenticity and integrity of messages or documents.

- **Secure storage and retrieval:** Employed in blockchain technology to create secure links between blocks of data.

Common cryptographic hashing algorithms include MD5 (Message Digest Algorithm 5), SHA-1 (Secure Hash Algorithm 1), SHA-256 (part of the Secure Hash Algorithm 2 family), and SHA-3. It's important to note that as computing power increases and new cryptographic vulnerabilities are discovered, older hashing algorithms may become less secure, and it's advisable to use the most up-to-date and robust hashing functions available.
