# File I/O


*   Files are sequence of lines
*   Each line is a String
*   Various access modes are used to define
how a file has to be used once it is **open**

## File Open


*   In order to read, write, or append to the file, first, it needs to be opened using the open() function.
*   The open() function has almost 8 parameters and among them, 2 are most important: **filename** and **access mode**

Now let's see how open() function works.


```
var_name = open('filename (with location)', mode = 'Access_Mode') # open file

.....perform operations.....

var_name.close() # close file (frees memory)
```
There are many access modes in Python and for this course, we will be only discussing the most
used ones.


*   **Read Only** (`'r'`): Opens a text file for reading from the start of the file. *If the file
does not exist or the given path is wrong, it raises an error.*
*   **Write Only** (`'w'`): Opens a text file for writing from the start of the file. *If the file does not exist, then creates a new file. If the file already exists, then overwrites its contents.*
*   **Append** (`'a'`): Opens a text file for writing from the end of the file. *If the file
does not exist, then create a new file. If the file already exists, then adds the new contents at the end of the old contents.*









### Reading from a File

Consider we have a file named **"hello.txt"** that has the following contents:


```
Hello World!
How are you?
I am learning CSE110.
Python is too easy!

```
We need to use `mode = 'r'` to read a file.

#### read() method

````python
f = open('hello.txt', mode = 'r')
contents = f.read()
print(type(contents))
print()
print(contents)
# print(f.read())
f.close()
````

````python
# Let's say your file is in a different folder/directory from your code file
# Then you need to provide the file location as well

f = open('/content/My Folder/New/hello1.txt', mode = 'r')
print(f.read())
f.close()
````

````python
# read() method takes an optional parameter, n which is the number of characters
# if n is not mentioned, then by default the whole text is printed

f = open('hello.txt', mode = 'r')
print(f.read(5)) # n = 5, which means I want to read the first 5 characters
f.close()
````

#### readline() method

````python
# readline() method prints only one line at a time

f = open('hello.txt', mode = 'r')
line1 = f.readline()
print(type(line1))
print(line1)
f.close()
````

````python
f = open('hello.txt', mode = 'r')
print(f.readline())
print(f.readline())
print(f.readline())
print(f.readline())
f.close()
````

````python
# While printing, readline() method ends with a new line (\n) by default
# you can avoid this by controlling the end parameter inside print

f = open('hello.txt', mode = 'r')
print(f.readline(),end = '')
print(f.readline(),end = '')
print(f.readline(),end = '')
print(f.readline(),end = '')
f.close()
````

````python
# or using rstrip() method

f = open('hello.txt', mode = 'r')
print(f.readline().rstrip())
print(f.readline().rstrip())
print(f.readline().rstrip())
print(f.readline().rstrip())
f.close()
````

````python
# like read() method, readline() also takes an optional parameter, n
# which is the number of characters
# if n is not mentioned, then by default the whole line's text is printed

f = open('hello.txt', mode = 'r')
print(f.readline(5))
f.close()
````

#### readlines() method

````python
# this method returns a list with each line of the file as its elements

f = open('hello.txt', mode = 'r')
myList = f.readlines()
print(type(myList))
print(myList)
f.close()
````

````python
f = open('hello.txt', mode = 'r')
print(f.readlines())
f.close()
````

````python
# readlines() method is easier for iteration

f = open('hello.txt', mode = 'r')
myList = f.readlines()
for idx in range(len(myList)):
    print(myList[idx])
f.close()
````

````python
# removing the blank lines after each line

f = open('hello.txt', mode = 'r')
myList = f.readlines()

# approach 1
for idx in range(len(myList)):
    print(myList[idx], end = "")

print()

# approach 2
for idx in range(len(myList)):
    print(myList[idx].rstrip())

f.close()
````

### Writing to a File

**WARNING:**
*   If the file doesn't exist, then using `mode = 'w'` creates a new file.
*  Otherwise, it **clears and overwrites** the current file.


#### write(s) method

````python
# this method write a single string (s) at the end of the file

f = open('new.txt', mode = 'w')
myStr = 'This is a line\n'
f.write(myStr)
f.close()

# now let's read the file
f = open('new.txt', mode = 'r')
print(f.read())
f.close()
````

````python
# to write multiple lines, use \n after the end of each line

f = open('new.txt', mode = 'w')
myStr = 'This is first line\nThis is second line\nThis is third line\n'
f.write(myStr)
f.close()

f = open('new.txt', mode = 'r')
print(f.read())
f.close()
````

#### writelines(S) method

````python
# here, the parameter S is a list of strings
# this method writes each element of S as a separate line to file

f = open('new2.txt', mode = 'w')
str1 = "This is first line\n"
str2 = "This is second line\n"
str3 = "This is third line\n"
f.writelines([str1,str2,str3])
f.close()

f = open('new2.txt', mode = 'r')
print(f.read())
f.close()
````

### Appending to a File

*   For appending, we open the file in append mode  `mode = 'a'`.
*  It is almost similar to 'w' mode. If the file does not exist, then a new file is created. We use the write(s) and writelines(S) methods to **append additional lines** to the file.


````python
# appending when file doesn't exist

f = open('new3.txt', mode = 'a')
line1 = "This is a line\n"
f.write(line1)
f.close()

f = open('new3.txt', mode = 'r')
print(f.read())
f.close()
````

````python
# using write(s) method

f = open('new3.txt', mode = 'r')
print(f.read())
f.close()

f = open('new3.txt', mode = 'a')
str1 = "This is another line\n"
f.write(str1)
f.close()

f = open('new3.txt', mode = 'r')
print(f.read())
f.close()
````

````python
# using writelines(S) method

f = open('new3.txt', mode = 'r')
print(f.read())
f.close()

f = open('new3.txt', mode = 'a')
str1 = "This is line 3\n"
str2 = "This is line 4\n"
f.writelines([str1,str2])
f.close()

print("------------------")
f = open('new3.txt', mode = 'r')
print(f.read())
f.close()
````
